#1. Análise do faturamento


```python


dataframes = { 'loja': loja, 'loja2': loja2, 'loja3': loja3, 'loja4': loja4}
coluna_preco = 'Preço'

faturamento_por_loja = {}

for nome_df, df in dataframes.items():
    if coluna_preco in df.columns:
        faturamento_total = df[coluna_preco].sum()
        faturamento_por_loja[nome_df] = faturamento_total
        print(f"O faturamento total da loja '{nome_df}' é: R$ {faturamento_total:.2f}")
    else:
        print(f"A coluna '{coluna_preco}' não existe no DataFrame '{nome_df}'.")

#print("\n--- Faturamento Total por Loja (em dicionário) ---")
#print(faturamento_por_loja)
```

```python



coluna_preco = 'Preço'
faturamento_sem_outliers = {}

for nome_df, df in dataframes.items():
    if coluna_preco in df.columns:
        Q1 = df[coluna_preco].quantile(0.25)
        Q3 = df[coluna_preco].quantile(0.75)
        IQR = Q3 - Q1

        limite_inferior_iqr = Q1 - 1.5 * IQR
        limite_superior_iqr = Q3 + 1.5 * IQR

        # Filtrando o DataFrame para remover os outliers
        df_sem_outliers = df[
            (df[coluna_preco] >= limite_inferior_iqr) & (df[coluna_preco] <= limite_superior_iqr)
        ].copy() # Use .copy() para evitar warnings ao modificar o DataFrame

        faturamento_total_sem_outliers = df_sem_outliers[coluna_preco].sum()
        faturamento_sem_outliers[nome_df] = faturamento_total_sem_outliers

        print(f"Faturamento total da loja '{nome_df}' (sem outliers IQR): R$ {faturamento_total_sem_outliers:.2f}")
    else:
        print(f"A coluna '{coluna_preco}' não existe no DataFrame '{nome_df}'.")

print("\n--- Faturamento Total por Loja (sem outliers IQR) ---")
print(faturamento_sem_outliers)
```

```python


coluna_preco = 'Preço'

faturamento_com_outliers = {}
faturamento_sem_outliers = {}

for nome_df, df in dataframes.items():
    if coluna_preco in df.columns:
        # Faturamento com outliers
        faturamento_com_outliers[nome_df] = df[coluna_preco].sum()

        # Identificação dos limites IQR
        Q1 = df[coluna_preco].quantile(0.25)
        Q3 = df[coluna_preco].quantile(0.75)
        IQR = Q3 - Q1
        limite_inferior_iqr = Q1 - 1.5 * IQR
        limite_superior_iqr = Q3 + 1.5 * IQR

        # Filtrando o DataFrame para remover os outliers
        df_sem_outliers = df[
            (df[coluna_preco] >= limite_inferior_iqr) & (df[coluna_preco] <= limite_superior_iqr)
        ][coluna_preco].sum()
        faturamento_sem_outliers[nome_df] = df_sem_outliers

    else:
        print(f"A coluna '{coluna_preco}' não existe no DataFrame '{nome_df}'.")

# Criando Series pandas para facilitar a ordenação
faturamento_com_outliers_series = pd.Series(faturamento_com_outliers).sort_values(ascending=False)
faturamento_sem_outliers_series = pd.Series(faturamento_sem_outliers).sort_values(ascending=False)

print("\n--- Faturamento Total por Loja (Com Outliers) ---")
print(faturamento_com_outliers_series.map(lambda x: f'R$ {x:.2f}'))

print("\n--- Faturamento Total por Loja (Sem Outliers IQR) ---")
print(faturamento_sem_outliers_series.map(lambda x: f'R$ {x:.2f}'))

print("\n--- Comparativo da Ordem das Lojas ---")
print("\nOrdem com Outliers:")
print(faturamento_com_outliers_series.index.tolist())

print("\nOrdem sem Outliers (IQR):")
print(faturamento_sem_outliers_series.index.tolist())

# Verificando se a ordem é a mesma
ordem_igual = faturamento_com_outliers_series.index.tolist() == faturamento_sem_outliers_series.index.tolist()
print(f"\nA ordem das lojas é a mesma com e sem outliers (IQR)? {ordem_igual}")
```

# 2. Vendas por Categoria


```python


for nome_df, df in dataframes.items():
    if 'Categoria do Produto' in df.columns:
        # Agrupa por categoria e conta as ocorrências
        contagem_por_categoria = df.groupby('Categoria do Produto').size().sort_values(ascending=False)

        print(f"\n--- Quantidade de Produtos Vendidos por Categoria na Loja '{nome_df}' ---")
        print(contagem_por_categoria)

        if not contagem_por_categoria.empty:
            categoria_mais_popular = contagem_por_categoria.index[0]
            quantidade_mais_popular = contagem_por_categoria.iloc[0]
            print(f"\nA categoria mais popular na loja '{nome_df}' é '{categoria_mais_popular}' com {quantidade_mais_popular} vendas.")
        else:
            print(f"Não há vendas registradas na loja '{nome_df}'.")
    else:
        print(f"\nA coluna 'Categoria do Produto' não existe no DataFrame '{nome_df}'.")
```

```python



coluna_preco = 'Preço'
coluna_categoria = 'Categoria do Produto'

for nome_df, df in dataframes.items():
    print(f"\n--- Análise da Loja '{nome_df}' ---")

    if coluna_preco in df.columns and coluna_categoria in df.columns:
        # --- Análise COM outliers ---
        contagem_por_categoria_com_outliers = df.groupby(coluna_categoria).size().sort_values(ascending=False)
        print("\n--- Quantidade de Produtos Vendidos por Categoria (Com Outliers) ---")
        print(contagem_por_categoria_com_outliers)
        if not contagem_por_categoria_com_outliers.empty:
            categoria_mais_popular_com_outliers = contagem_por_categoria_com_outliers.index[0]
            quantidade_mais_popular_com_outliers = contagem_por_categoria_com_outliers.iloc[0]
            print(f"Categoria mais popular (com outliers): '{categoria_mais_popular_com_outliers}' com {quantidade_mais_popular_com_outliers} vendas.")
        else:
            print("Não há vendas registradas.")

        # --- Identificação e Remoção de Outliers (usando IQR na coluna de preço) ---
        Q1 = df[coluna_preco].quantile(0.25)
        Q3 = df[coluna_preco].quantile(0.75)
        IQR = Q3 - Q1
        limite_inferior_iqr = Q1 - 1.5 * IQR
        limite_superior_iqr = Q3 + 1.5 * IQR
        df_sem_outliers = df[
            (df[coluna_preco] >= limite_inferior_iqr) & (df[coluna_preco] <= limite_superior_iqr)
        ].copy()

        # --- Análise SEM outliers ---
        if not df_sem_outliers.empty and coluna_categoria in df_sem_outliers.columns:
            contagem_por_categoria_sem_outliers = df_sem_outliers.groupby(coluna_categoria).size().sort_values(ascending=False)
            print("\n--- Quantidade de Produtos Vendidos por Categoria (Sem Outliers IQR) ---")
            print(contagem_por_categoria_sem_outliers)
            if not contagem_por_categoria_sem_outliers.empty:
                categoria_mais_popular_sem_outliers = contagem_por_categoria_sem_outliers.index[0]
                quantidade_mais_popular_sem_outliers = contagem_por_categoria_sem_outliers.iloc[0]
                print(f"Categoria mais popular (sem outliers IQR): '{categoria_mais_popular_sem_outliers}' com {quantidade_mais_popular_sem_outliers} vendas.")
            else:
                print("Não há vendas registradas após a remoção de outliers.")
        else:
            print("Não há dados disponíveis após a remoção de outliers para analisar categorias.")

    else:
        print(f"Certifique-se de que as colunas '{coluna_preco}' e '{coluna_categoria}' existem no DataFrame '{nome_df}'.")
```

# 3. Média de Avaliação das Lojas

```python
coluna_avaliacao = 'Avaliação da compra'
coluna_preco = 'Preço'  # Usada para identificar outliers

for nome_df, df in dataframes.items():
    print(f"\n--- Análise de Avaliações da Loja '{nome_df}' ---")

    if coluna_avaliacao in df.columns:
        # --- Média das Avaliações COM outliers ---
        media_avaliacao_com_outliers = df[coluna_avaliacao].mean()
        print(f"Média das avaliações dos clientes (com todos os dados): {media_avaliacao_com_outliers:.2f}")

        # --- Identificação e Remoção de Outliers (usando IQR na coluna de preço) ---
        if coluna_preco in df.columns:
            Q1_preco = df[coluna_preco].quantile(0.25)
            Q3_preco = df[coluna_preco].quantile(0.75)
            IQR_preco = Q3_preco - Q1_preco
            limite_inferior_iqr_preco = Q1_preco - 1.5 * IQR_preco
            limite_superior_iqr_preco = Q3_preco + 1.5 * IQR_preco
            df_sem_outliers_preco = df[
                (df[coluna_preco] >= limite_inferior_iqr_preco) & (df[coluna_preco] <= limite_superior_iqr_preco)
            ].copy()

            # --- Média das Avaliações SEM outliers (baseado em outliers de preço) ---
            if coluna_avaliacao in df_sem_outliers_preco.columns and not df_sem_outliers_preco.empty:
                media_avaliacao_sem_outliers_preco = df_sem_outliers_preco[coluna_avaliacao].mean()
                print(f"Média das avaliações dos clientes (sem outliers de preço IQR): {media_avaliacao_sem_outliers_preco:.2f}")
            else:
                print("Não há dados de avaliação disponíveis após a remoção de outliers de preço.")
        else:
            print("A coluna 'Preço' não encontrada, impossível remover outliers baseados no preço para análise de avaliação.")

    else:
        print(f"A coluna '{coluna_avaliacao}' não existe no DataFrame '{nome_df}'.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns


df = lojas.copy()  # Criando uma cópia para não modificar o original

coluna_preco = 'Preço'
coluna_loja = 'lojas'

# --- Faturamento Total por Loja (COM outliers) ---
faturamento_por_loja_com_outliers = df.groupby(coluna_loja)[coluna_preco].sum().sort_values(ascending=False)
print("\n--- Faturamento Total por Loja (Com Outliers) ---")
print(faturamento_por_loja_com_outliers)

# Visualização (Com Outliers)
plt.figure(figsize=(10, 6))
sns.barplot(x=faturamento_por_loja_com_outliers.index, y=faturamento_por_loja_com_outliers.values, palette='viridis')
plt.title('Faturamento Total por Loja (Com Outliers)')
plt.xlabel('Loja')
plt.ylabel('Faturamento Total (R$)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# --- Identificação e Remoção de Outliers (usando IQR na coluna de preço) ---
Q1_preco = df[coluna_preco].quantile(0.25)
Q3_preco = df[coluna_preco].quantile(0.75)
IQR_preco = Q3_preco - Q1_preco
limite_inferior_iqr_preco = Q1_preco - 1.5 * IQR_preco
limite_superior_iqr_preco = Q3_preco + 1.5 * IQR_preco
df_sem_outliers_preco = df[
    (df[coluna_preco] >= limite_inferior_iqr_preco) & (df[coluna_preco] <= limite_superior_iqr_preco)
].copy()

# --- Faturamento Total por Loja (SEM outliers) ---
faturamento_por_loja_sem_outliers = df_sem_outliers_preco.groupby(coluna_loja)[coluna_preco].sum().sort_values(ascending=False)
print("\n--- Faturamento Total por Loja (Sem Outliers IQR) ---")
print(faturamento_por_loja_sem_outliers)

# Visualização (Sem Outliers)
plt.figure(figsize=(10, 6))
sns.barplot(x=faturamento_por_loja_sem_outliers.index, y=faturamento_por_loja_sem_outliers.values, palette='magma')
plt.title('Faturamento Total por Loja (Sem Outliers de Preço IQR)')
plt.xlabel('Loja')
plt.ylabel('Faturamento Total (R$)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# --- Comparativo Lado a Lado ---
faturamento_comparativo = pd.DataFrame({
    'Com Outliers': faturamento_por_loja_com_outliers,
    'Sem Outliers (IQR)': faturamento_por_loja_sem_outliers
})

faturamento_comparativo.plot
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Supondo que 'lojas' seja o seu DataFrame consolidado
df = lojas.copy()

coluna_produto = 'Produto'
coluna_loja = 'lojas'
coluna_preco = 'Preço' # Para remoção de outliers
n_top_bottom = 5

def plot_top_bottom_products_barh(df_loja, n=5, title_suffix=''):
    contagem = df_loja[coluna_produto].value_counts().sort_values(ascending=False)
    top_n = contagem.head(n)
    bottom_n = contagem.tail(n)

    plt.figure(figsize=(10, 8))

    plt.subplot(2, 1, 1)
    sns.barplot(x=top_n.values, y=top_n.index, palette='Greens_r')
    plt.title(f'Top {n} Produtos Mais Vendidos - Loja {df_loja[coluna_loja].iloc[0]}{title_suffix}')
    plt.xlabel('Quantidade Vendida')
    plt.ylabel('Produto')

    plt.subplot(2, 1, 2)
    sns.barplot(x=bottom_n.values, y=bottom_n.index, palette='Reds_r')
    plt.title(f'Top {n} Produtos Menos Vendidos - Loja {df_loja[coluna_loja].iloc[0]}{title_suffix}')
    plt.xlabel('Quantidade Vendida')
    plt.ylabel('Produto')

    plt.tight_layout()
    plt.show()

for loja_nome in df[coluna_loja].unique():
    df_loja = df[df[coluna_loja] == loja_nome].copy()
    plot_top_bottom_products_barh(df_loja, n=n_top_bottom, title_suffix=' (Com Outliers)')

    # Remoção de outliers (baseado no preço)
    Q1_preco = df_loja[coluna_preco].quantile(0.25)
    Q3_preco = df_loja[coluna_preco].quantile(0.75)
    IQR_preco = Q3_preco - Q1_preco
    limite_inferior_iqr_preco = Q1_preco - 1.5 * IQR_preco
    limite_superior_iqr_preco = Q3_preco + 1.5 * IQR_preco
    df_sem_outliers_preco_loja = df_loja[
        (df_loja[coluna_preco] >= limite_inferior_iqr_preco) & (df_loja[coluna_preco] <= limite_superior_iqr_preco)
    ].copy()

    if not df_sem_outliers_preco_loja.empty:
        plot_top_bottom_products_barh(df_sem_outliers_preco_loja, n=n_top_bottom, title_suffix=' (Sem Outliers Preço IQR)')
    else:
        print(f"\nNão há dados suficientes após remoção de outliers de preço para loja {loja_nome}.")
```
